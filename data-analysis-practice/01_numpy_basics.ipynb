## 1. Creating NumPy Arrays and Array Information

```python
import numpy as np
```

```python
numbers = np.array([10, 20, 30, 40, 50])
print(numbers)
print(numbers.shape)
print(numbers.size)
print(numbers.dtype)
```

```python
marks = np.array([65, 70, 85, 90, 55])
print(marks)
print(marks.shape)
print(marks.size)
print(marks.dtype)
```

```python
matrix = np.array([[1,2,3],[3,2,1]])
print(matrix)
print(matrix.shape)
print(matrix.size)
print(matrix.dtype)
```

## 2. Indexing

```python
num = np.array([10, 20, 30, 40, 50])
print(num[0])
print(num[2])
print(num[4])
```

```python
matrix = np.array([[1,2,3],[3,2,1]])
print(matrix[0][0])
print(matrix[1][2])
```

## 3. Slicing

```python
num = np.array([10, 20, 30, 40, 50])
print(num[0:3])
```

```python
num = np.array([10, 20, 30, 40, 50])
print(num[1:4])
```

```python
num = np.array([10, 20, 30, 40, 50])
print(num[2:])
```

```python
num = np.array([10, 20, 30, 40, 50])
print(num[3:])
```

```python
num = np.array([10, 20, 30, 40, 50])
print(num[:3])
```

```python
num = np.array([10, 20, 30, 40, 50])
print(num[:])
```

## 4. Statistical Functions

```python
marks = np.array([65, 70, 85, 90, 55])
```

```python
print(sum(marks))
```

```python
print(max(marks))
```

```python
print(min(marks))
```

```python
print(np.mean(marks))
```

```python
print(np.std(marks))
```

```python
print(np.median(marks))
```

## 5. Reshape

```python
num = np.array([1, 2, 3, 4, 5, 6])
```

```python
new_2D = num.reshape(2,3)
print(new_2D)
```

```python
new_3D = num.reshape(3,2)
print(new_3D)
```

## 6. Power Functions

```python
square = np.square(5)
print(square)
```

```python
sqrt = np.sqrt(25)
print(sqrt)
```

```python
pow = np.pow(5,2)
print(pow)
```
