```python
class BankAccount:
    def __init__(self,account_holder,account_number,account_balance,account_password):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__account_balance = account_balance
        self.__account_password = account_password

    def show_account_information(self):
        print(f"Account holder name : {self.account_holder}\nAccount number : {self.account_number}")

    def show_account_balance(self):
        print(f"Account balance : {self.__account_balance}")

    def deposit_balance(self,deposit_amount):
        if(deposit_amount > 0):
            self.__account_balance = self.__account_balance + deposit_amount
        else:
            print("Please enter deposit amount in positive.")

    def withdraw_balance(self,withdraw_amount):
        
        if(withdraw_amount <= 0):
            print("Please enter withdraw amount in positive.")
        elif(self.__account_balance < withdraw_amount):
            print("Insufficient balance.")    
        else:
            self.__account_balance = self.__account_balance - withdraw_amount      

    def password_change(self,old_password,new_password):
        if(self.__account_password == old_password):
            self.__account_password = new_password
            print("Password changed successfully.")
        else:
            print("Old password not match.")             


bankaccount = BankAccount("Ali Raza",14137,100000,"1413")
while(True):
    try:
        print("Please enter your option.\n1 for show account information\n2 for show balance\n3 for deposit\n4 for withdraw\n5 for password change\n6 for exit\n")
        option = int(input("Please enter your option number : "))

    except ValueError:
        print("Please enter only number value between 1 to 6.")

    else:
        if(option == 1):
            bankaccount.show_account_information() 
        elif(option == 2):
            bankaccount.show_account_balance()
        elif(option == 3):
            try:
                deposit = int(input("Please enter deposit amount : "))
            except ValueError:
                print("Please enter deposit amount in integer.")
            else:
                bankaccount.deposit_balance(deposit)        

        elif(option == 4):
            try:
                withdraw = int(input("Please enter withdraw amount : "))
            except ValueError:
                print("Please enter withdraw amount in integer..")
            else:
                bankaccount.withdraw_balance(withdraw) 
        elif(option == 5):
            old_password = input("Please enter old password : ")
            new_password = input("Please enter new password : ")
            if(len(old_password) == 4 and old_password.isdigit()):
                if(len(new_password) == 4 and new_password.isdigit()):
                    bankaccount.password_change(old_password,new_password)
                else:
                    print("Please enter new password in digit and length should be 4 didt.")    
            else:
                print("Please enter old password in digit and length should be 4 didt.")         
        elif(option == 6):
            break             
        else:
            print("Please enter option number between 1 to 6.")    
```
