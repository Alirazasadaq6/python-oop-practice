```python
class Teacher:
    def __init__(self,subject):
        self.subject = subject

    def show_subject(self):
        print(f"Subject : {self.subject}")    

class Researcher:
    def __init__(self,research_area):
        self.research_area = research_area

    def show_research_area(self):
        print(f"Research area : {self.research_area}")    

class Professor(Teacher,Researcher):
    def __init__(self,subject,research_area):
        Teacher.__init__(self,subject)
        Researcher.__init__(self,research_area)


professor = Professor("Computing","Machine Learning")
professor.show_subject()
professor.show_research_area()
```
