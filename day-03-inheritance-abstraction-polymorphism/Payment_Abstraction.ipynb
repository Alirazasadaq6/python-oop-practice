```python
from abc import ABC, abstractmethod
class Payment(ABC):
    @abstractmethod
    def pay(self):
        pass
class CardPayment(Payment):
    def pay(self):
        print(f"Payment made using Card")

class PayPalPayment(Payment):
    def pay(self):
        print(f"Payment made using PayPal")

class BankTransfer(Payment):
    def pay(self):
        print(f"Payment made using Bank Transfer")

cardpayment = CardPayment()
paypalpayment = PayPalPayment()
banktransfer = BankTransfer()

cardpayment.pay()
paypalpayment.pay()
banktransfer.pay()
```
