```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age =age

    def display_info(self):
        print(f"Name : {self.name}\nAge : {self.age}")

class Student(Person):
    def __init__(self,course,name,age):
        super().__init__(name,age)
        self.course = course

    def show_course(self):
        print(f"Student Course : {self.course}")    

class Teacher(Person):
    def __init__(self,subject,name,age):
        super().__init__(name,age)
        self.subject = subject

    def show_subject(self):
        print(f"Teacher Subject : {self.subject}")

student = Student("Computing","Ali",26)
teacher = Teacher("OOP","Raza",35)   

student.display_info()
student.show_course()

teacher.display_info()
teacher.show_subject()
```
