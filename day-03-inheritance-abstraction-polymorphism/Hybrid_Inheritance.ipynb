```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age =age

class Student(Person):
    def __init__(self,course,name,age):
        Person.__init__(self,name,age)
        self.course = course  

class Teacher(Person):
    def __init__(self,subject,name,age):
        Person.__init__(self,name,age)
        self.subject = subject

class Assistant(Student,Teacher):
    def __init__(self,stipend,name,age,course,subject):
        Student.__init__(self,course,name,age)
        Teacher.__init__(self,subject,name,age)
        self.stipend = stipend

    def show_assistant_info(self):
        print(f"Name : {self.name}\nAge : {self.age}\nCourse : {self.course}\nSubject : {self.subject}\nStipend : {self.stipend}")

assistant = Assistant(25_000,"AliRaza",26,"Computing","OOP")
assistant.show_assistant_info()
```
