```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def __init__(self,dog_sound):
        self.dog_sound = dog_sound
    def sound(self):
        print(f"Dog Sound : {self.dog_sound}")

class Cat(Animal):
    def __init__(self,cat_sound):
        self.cat_sound = cat_sound
    def sound(self):
        print(f"Cat Sound : {self.cat_sound}")

dog = Dog("Bark")
dog.sound()

cat = Cat("Meow")
cat.sound()
```
