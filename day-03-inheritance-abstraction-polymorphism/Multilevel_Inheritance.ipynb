```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name : {self.name}\nAge : {self.age}")

class Student(Person):
    def __init__(self,student_id,course,grade,name,age):
        super().__init__(name,age)
        self.student_id = student_id
        self.course = course
        self.grade = grade

    def show_course(self):
        print(f"Student ID : {self.student_id}\nCourse : {self.course}\nGrade : {self.grade}")

class GraduateStudent(Student):
    def __init__(self, thesis_topic, student_id, course, grade, name, age):
        super().__init__(student_id,course,grade,name,age)
        self.thesis_topic = thesis_topic

    def show_thesis(self):
        print(f"Thesis Topic : {self.thesis_topic}")    

graduatestudent = GraduateStudent("Animal Detection",12345,"Computing","A","AliRaza",27)
graduatestudent.display_info()
graduatestudent.show_course()
graduatestudent.show_thesis()
```
