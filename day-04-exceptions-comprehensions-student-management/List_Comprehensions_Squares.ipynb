## Squares

```python
number = [i*i for i in range(6)]

print(number)
```

## Even Numbers

```python
even_number = [i for i in range(10) if i%2 == 0]

print(even_number)
```

## Odd Numbers

```python
odd_number = [i for i in range(10) if i%2 != 0]

print(odd_number)
```

## Filter Names

```python
name = ['Ali','Ahmed','Sara','Ayesha','John']

filter_name = [i for i in name if i[0] == 'A']
print(filter_name)
```

## Uppercase Conversion

```python
name = ['ali','ahmed','sara','ayesha','john']

upper_case = [i.upper() for i in name]
print(upper_case)
```
