```python
import json
class Students:
    def __init__(self):
        pass

    def view_record(self):
        with open("students.json","r") as file:
            student_view_record = json.load(file)
            for svr in student_view_record:
                print(f"{svr["id"]} - {svr["name"]} - {svr["department"]} - ({svr["subjects"][0]},{svr["subjects"][1]})")

    def add_record(self,name,department,subject_1,subject_2):
         student_id = 1001
         with open("students.json","r") as file:
            student_add_record = json.load(file)
            if len(student_add_record) != 0:
                student_id = student_add_record[-1]["id"] + 1

            new_student = {
                "id": student_id,
                "name" : name,
                "department" : department,
                "subjects": [subject_1, subject_2]
            }
            student_add_record.append(new_student)
            with open("students.json","w") as file:
                json.dump(student_add_record,file,indent=4)
                print("Student added successfully.")

    def search_record(self,student_id):
        found = False
        with open("students.json","r") as file:
            student_search_record = json.load(file)
            for ssr in student_search_record:
                if(ssr["id"] == student_id):
                    print(f"{ssr["id"]} - {ssr["name"]} - {ssr["department"]} - ({ssr["subjects"][0]},{ssr["subjects"][1]})")
                    found = True
                    break

            if found == False:
                print("Student not found.")

    def update_record(self,student_id):
        found = False
        update = False
        record_index = -1
        with open("students.json","r") as file:
            student_update_record = json.load(file)
            for index, sur in enumerate(student_update_record):
                if(sur["id"] == student_id):
                    print(f"{sur["id"]} - {sur["name"]} - {sur["department"]} - ({sur["subjects"][0]},{sur["subjects"][1]})")
                    found = True
                    student_id = sur["id"]
                    student_name = sur["name"]
                    student_department = sur["department"]
                    student_subject_1 = sur["subjects"][0]
                    student_subject_2 = sur["subjects"][1]
                    record_index = index
                    break

            if found == False:
                print("Student not found.")
            else:
                try:
                    print("Please selecte your option.\n1 for name update\n2 for department update.\n3 for subject 1 update.\n4 for subject 2 update.")
                    update_option = int(input("Please enter your option : "))
                except ValueError:
                    print("Please enter only integer value.")
                else:
                    if(update_option == 1):
                        update_name = input("Please enter update name : ")
                        student_name = update_name
                        update = True

                    elif(update_option == 2):
                        update_department = input("Please enter update department : ")
                        student_department = update_department
                        update = True

                    elif(update_option == 3):
                        update_subject_1 = input("Please enter update subject 1 : ")
                        student_subject_1 = update_subject_1
                        update = True

                    elif(update_option == 4):
                        update_subject_2 = input("Please enter update subject 2 : ")
                        student_subject_2 = update_subject_2
                        update = True

                    else:
                        print("Please enter your option between 1 to 4.")

                    if update == True:
                        update_student = {
                            "id": student_id,
                            "name" : student_name,
                            "department" : student_department,
                            "subjects": [student_subject_1, student_subject_2]
                        }
                        student_update_record[record_index] = update_student
                        with open("students.json","w") as file:
                            json.dump(student_update_record,file,indent=4)
                            print("Student updated successfully.")

    def delete_record(self,student_id):
        found = False
        record_index = -1
        with open("students.json","r") as file:
            student_delete_record = json.load(file)
            for index, sdr in enumerate(student_delete_record):
                if(sdr["id"] == student_id):
                    print(f"{sdr["id"]} - {sdr["name"]} - {sdr["department"]} - ({sdr["subjects"][0]},{sdr["subjects"][1]})")
                    found = True
                    record_index = index
                    break

            if found == False:
                print("Student not found.")
            else:
                del student_delete_record[record_index]
                with open("students.json","w") as file:
                    json.dump(student_delete_record,file,indent=4)
                    print("Student deleted successfully.")

students = Students()
while(True):
    try:
        print("Please select option.\n1 for view record.\n2 for add record.\n3 for search record.\n4 for update record.\n5 for delete record.\n6 for exit.")
        option = int(input("Please enter your option number : "))

    except ValueError:
        print("Please enter only integer value.")

    else:
        if(option == 1):
            students.view_record()

        elif(option == 2):
            student_name = input("Please enter student name : ")
            student_department = input("Please enter student department name : ")
            student_subject_1 = input("Please enter student subject 1 : ")
            student_subject_2 = input("Please enter student subject 2 : ")

            students.add_record(student_name,student_department,student_subject_1,student_subject_2)

        elif(option == 3):
            student_id = input("Please enter student id : ")
            if(len(student_id) == 4 and student_id.isdigit()):
                students.search_record(int(student_id))
            else:
                print("Please enter student id in digit and length should be 4 digit.")  

        elif(option == 4):
            student_id = input("Please enter student id : ")
            if(len(student_id) == 4 and student_id.isdigit()):
                students.update_record(int(student_id))
            else:
                print("Please enter student id in digit and length should be 4 digit.") 

        elif(option == 5):            
            student_id = input("Please enter student id : ")
            if(len(student_id) == 4 and student_id.isdigit()):
                students.delete_record(int(student_id))
            else:
                print("Please enter student id in digit and length should be 4 digit.")  

        elif(option == 6):
            print("Thank You!!!")
            break
        else:
            print("Please enter your option between 1 to 6.")
```
