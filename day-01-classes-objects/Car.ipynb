```python
class Car:
    def __init__(self,brand,model,speed,fuel,is_started):
        self.brand = brand
        self.model = model
        self.speed = speed
        self.fuel = fuel
        self.is_started = is_started

    def show_details(self):
        if(self.is_started == 1):
            car_start = "Yes"
        else:
            car_start = "No"    
        print(f"Car Brand : {self.brand}\nCar Model : {self.model}\nCar Speed : {self.speed}\nFuel in Car : {self.fuel}\nCar is Started : {car_start}")

    def start_car(self):
        if(self.is_started == 1):
            print("Car already start")
        else:
            if(self.fuel > 0):
                print("Car is started.")
                self.is_started = 1
            else:
                print("Insufficient fuel")

    def accelerate(self):
        if(self.is_started == 0):
            print("Car is not started.")
        else:
            if(self.fuel > 0):
                self.speed = self.speed + 10
                self.fuel = self.fuel - 1
                print(f"Speed increased, new speed is {self.speed}")
            else:
                print("Insufficient fuel")

    def brake(self):
        if(self.is_started == 0):
            print("Car is not started.")
        elif(self.speed >= 5):
            self.speed = self.speed - 5
            print(f"Speed decreased, new speed is {self.speed}")
        elif(self.speed > 0 and self.speed < 5):
            self.speed = 0
            print(f"Speed decreased, now speed is 0")
        else:
            print(f"Your car speed already zero.")

    def refuel(self):
        try:
           fuel_q = int(input("Please enter fuel quanity : "))
        except ValueError:
           print("Please fuel quanity between 0 to 10.")
        else:
            temp = fuel_q + self.fuel
            if(fuel_q <= 0):
                print("Please fuel amount in positive.")
            elif(temp > 100):
                print(f"You enter a overflue value, already fuel in your {self.fuel}, and your just add {100 - self.fuel}, so please enter a fule amount correct.") 
            else:
                self.fuel = self.fuel + fuel_q
                print(f"Refuel doe and now fuel is {self.fuel}")    

car = Car("Teyota", 2025, 0, 0, 0)

while(True):

    try:
        print(f"Please select your option.\n1 for car details.\n2 for car start.\n3 for accelerate.\n4 for brake.\n5 for refuel\n6 for exit.")
        option = int(input("Please enter your option : "))
    except ValueError:
        print("Please enter your option in integer.")
    else:
        if(option == 1):
            car.show_details()
        elif(option == 2):
            car.start_car()
        elif(option == 3):
            car.accelerate()
        elif(option == 4):
            car.brake()
        elif(option == 5):
            car.refuel()
        elif(option == 6):
            break
        else:
            print("Please enter your option between 1 to 6.")                      
            
        
```
