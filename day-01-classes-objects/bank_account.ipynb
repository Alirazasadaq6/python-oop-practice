```python
class BankAccount:
    def __init__(self,balance):
        self.balance = balance

    def show_balance(self):
        print(f"Your balance is : {self.balance}")

    def deposit_amount(self,d_balance):
        self.balance = self.balance + d_balance
        print(f"Your balance has updated, new balance is : {self.balance}")
        return self.balance

    def withdraw_amount(self,w_balance):
        self.balance = self.balance - w_balance
        print(f"Your balance has updated, new balance is : {self.balance}")
        return self.balance

bankaccount = BankAccount(10000)
while(True):
    try:
        print("Please Select Option.\n1 For Check balance\n2 For Deposit\n3 For Withdraw\n4 For exit\n")
        option = int(input("Please enter your option."))
    except ValueError:
        print("Please enter only integer number.")

    else:    
        if(option == 1):
            bankaccount.show_balance()

        elif(option == 2):
            try:
                d_balance = int(input("Please enter your deposit amount : "))
            except ValueError:
                print("Please enter only integer number.")
            else:
                if(d_balance <= 0):
                    print("Please add balance greater then 0.")
                else:
                    bankaccount.deposit_amount(d_balance)

        elif(option == 3):
            try:
                w_balance = int(input("Please enter withdraw amount : "))
            except ValueError:
                print("Please enter only integer number.")
            else:    
                if w_balance <= 0:
                    print("Please enter amount greater than 0.")
                elif w_balance > bankaccount.balance:
                    print("Insufficient balance.")
                else:
                    bankaccount.withdraw_amount(w_balance)  

        elif(option == 4):
            print("Thank you for using our banking system.")
            break 

        else:
            print("Please select option between 1 to 4.")
```
